**Wholesale Customer Spending EDA**(**HCL Project**)

Program #1: Import Libraries and Load Dataset

```python
# Program #1: Load the dataset

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Read the CSV file
data = pd.read_csv("Wholesale customers data.csv")

# Display first five records
print("First 5 records:")
print(data.head())
```

Program #2: Dataset Overview

```python
# Program #2: Dataset Overview

# Display last 5 records
print("Last 5 records:")
print(data.tail())

# Display number of rows and columns
print("\nDataset Shape:")
print(data.shape)

# Display information about the dataset
print("\nDataset Information:")
data.info()

# Display statistical summary
print("\nStatistical Summary:")
print(data.describe())
```

Program #3: Count Customers in Each Channel and Region

```python
# Program #3: Customer Count by Channel and Region

# Count customers in each channel
print("Number of customers in each Channel:")
print(data["Channel"].value_counts().sort_index())

# Count customers in each region
print("\nNumber of customers in each Region:")
print(data["Region"].value_counts().sort_index())
```

Program #4: Highest Total and Average Spending Category

```python
# Program #4: Highest Total and Average Spending Category

# Select product categories
categories = ["Fresh", "Milk", "Grocery", "Frozen",
              "Detergents_Paper", "Delicassen"]

# Calculate total spending
total_spending = data[categories].sum()

# Calculate average spending
average_spending = data[categories].mean()

print("Total Spending in Each Category:")
print(total_spending)

print("\nAverage Spending in Each Category:")
print(average_spending)

# Find highest total and average
print("\nCategory with Highest Total Spending:")
print(total_spending.idxmax())

print("\nCategory with Highest Average Spending:")
print(average_spending.idxmax())
```

Program #5: Statistical Analysis Using NumPy

```python
# Program #5: Statistical Analysis Using NumPy

categories = ["Fresh", "Milk", "Grocery", "Frozen",
              "Detergents_Paper", "Delicassen"]

# Convert category data into NumPy array
array = data[categories].to_numpy()

print("Statistical Analysis of Product Categories:\n")

for i, category in enumerate(categories):
    print(category)
    print("Mean:", np.mean(array[:, i]))
    print("Median:", np.median(array[:, i]))
    print("Minimum:", np.min(array[:, i]))
    print("Maximum:", np.max(array[:, i]))
    print()
```

Program #6: Find the Top 10 Highest-Spending Customers

```python
# Program #6: Top 10 Highest-Spending Customers

categories = ["Fresh", "Milk", "Grocery", "Frozen",
              "Detergents_Paper", "Delicassen"]

# Calculate total spending of each customer
data["Total_Spending"] = data[categories].sum(axis=1)

# Sort customers by total spending
top_10 = data.sort_values(
    by="Total_Spending", ascending=False
).head(10)

print("Top 10 Highest-Spending Customers:")
print(top_10[["Total_Spending"]])
```

Program #7: Compare Average Spending Between Two Channels

```python
# Program #7: Average Spending Comparison Between Channels

categories = ["Fresh", "Milk", "Grocery", "Frozen",
              "Detergents_Paper", "Delicassen"]

# Calculate average spending for each channel
channel_average = data.groupby("Channel")[categories].mean()

print("Average Spending Between Two Channels:")
print(channel_average)
```

Program #8: Check Missing Values and Duplicate Records

```python
# Program #8: Missing Values and Duplicate Records

# Check missing values
print("Missing Values in Each Column:")
print(data.isnull().sum())

# Count total missing values
print("\nTotal Missing Values:")
print(data.isnull().sum().sum())

# Check duplicate records
print("\nTotal Duplicate Records:")
print(data.duplicated().sum())
```

Program #9 – Bar Chart of Customer Count by Channel.

```python
# Program #9: Channel Count Bar Chart

# Count customers in each channel
channel_counts = data["Channel"].value_counts().sort_index()

# Create bar chart
plt.bar(channel_counts.index, channel_counts.values)

# Add chart title and labels
plt.title("Number of Customers in Each Channel")
plt.xlabel("Channel")
plt.ylabel("Number of Customers")

# Display chart
plt.show()
```

Program #10 – Product Category Spending Bar Chart.

```python
# Program #10: Product Category Spending Bar Chart

# Select product categories
categories = ["Fresh", "Milk", "Grocery", "Frozen",
              "Detergents_Paper", "Delicassen"]

# Calculate total spending
category_spending = data[categories].sum()

# Create bar chart
plt.figure(figsize=(10, 5))
plt.bar(category_spending.index, category_spending.values)

# Add chart title and labels
plt.title("Total Spending by Product Category")
plt.xlabel("Product Categories")
plt.ylabel("Total Spending")
plt.xticks(rotation=30)

# Display chart
plt.show()
```

**Observation 2:** Fresh has the highest total spending among the six product categories. Delicassen has the lowest total spending.

**Observation 3:** The average spending pattern differs between the two channels. Channel 1 customers generally spend more on Fresh, Frozen and Delicassen, while Channel 2 customers generally spend more on Milk, Grocery and Detergents_Paper.
